```python
import pandas as pd
import numpy as np
```

## 1. DataFrame with 3 columns and 5 rows; add product of first two columns

```python
data = np.random.randint(1, 10, size=(5, 3))
df = pd.DataFrame(data, columns=['A', 'B', 'C'])

df['Product'] = df['A'] * df['B']
print(df)
```

## 2. Row-wise and column-wise sum

```python
data = np.random.randint(1, 10, size=(4, 3))
df = pd.DataFrame(data, columns=['A', 'B', 'C'])

print("DataFrame:")
print(df)
print("\nRow-wise sum:")
print(df.sum(axis=1))
print("\nColumn-wise sum:")
print(df.sum(axis=0))
```

## 3. Fill NaN values with the mean of their respective columns

```python
data = np.random.randint(1, 10, size=(5, 3)).astype(float)
df = pd.DataFrame(data, columns=['A', 'B', 'C'])

df.iloc[0, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Original DataFrame:")
print(df)
df = df.fillna(df.mean())
print("\nAfter filling NaN values:")
print(df)
```

## 4. Drop rows containing any NaN value

```python
data = np.random.randint(1, 10, size=(6, 4)).astype(float)
df = pd.DataFrame(data, columns=['A', 'B', 'C', 'D'])

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan

print("Original DataFrame:")
print(df)
df = df.dropna()
print("\nAfter dropping rows with NaN:")
print(df)
```

## 5. Group by Category and calculate sum and mean

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(10, 100, size=10)
})

print("Original DataFrame:")
print(df)
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("\nSum and mean for each category:")
print(result)
```

## 6. Group products by Category and calculate total sales

```python
df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Scale', 'Box'],
    'Category': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Sales': np.random.randint(100, 1000, size=6)
})

print("Original DataFrame:")
print(df)
result = df.groupby('Category')['Sales'].sum()
print("\nTotal sales for each category:")
print(result)
```

## 7. Merge two DataFrames using a common column

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Amit', 'Ravi', 'John']
})
df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 95]
})

result = pd.merge(df1, df2, on='ID')
print(result)
```

## 8. MultiIndex DataFrame: indexing and slicing

```python
arrays = [
    ['A', 'A', 'B', 'B'],
    ['One', 'Two', 'One', 'Two']
]
index = pd.MultiIndex.from_arrays(arrays, names=['Category', 'SubCategory'])

df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print("Original DataFrame:")
print(df)
print("\nIndexing Category A:")
print(df.loc['A'])
print("\nSlicing Categories A to B:")
print(df.loc['A':'B'])
```

## 9. MultiIndex with Category and SubCategory; calculate grouped sums

```python
df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B', 'A', 'B'],
    'SubCategory': ['X', 'Y', 'X', 'Y', 'X', 'Y'],
    'Value': np.random.randint(10, 100, size=6)
})

df = df.set_index(['Category', 'SubCategory'])
print("Original MultiIndex DataFrame:")
print(df)

result = df.groupby(level=['Category', 'SubCategory']).sum()
print("\nSum for each Category and SubCategory:")
print(result)
```

## 10. Apply a function to double all DataFrame values

```python
data = np.random.randint(1, 10, size=(5, 3))
df = pd.DataFrame(data, columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)
result = df.apply(lambda column: column * 2)
print("\nAfter doubling:")
print(result)
```

## 11. Use a lambda function to create a row-wise sum column

```python
data = np.random.randint(1, 10, size=(6, 3))
df = pd.DataFrame(data, columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
print(df)
```

## 12. Convert five text strings to uppercase

```python
s = pd.Series(['python', 'pandas', 'numpy', 'machine learning', 'data science'])

print("Original Series:")
print(s)
print("\nUppercase Series:")
print(s.str.upper())
```

## 13. Extract the first three characters of five strings

```python
s = pd.Series(['python', 'pandas', 'numpy', 'machine', 'science'])

print("Original Series:")
print(s)
print("\nFirst three characters:")
print(s.str[:3])
```
